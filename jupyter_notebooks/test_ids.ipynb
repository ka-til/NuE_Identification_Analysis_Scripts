```python
count = []
for file in filelistCC[48000:49000]:
    #print(file)
    infile=dataio.I3File(file)
    while(infile.more()):
        frame = infile.pop_frame()
        if frame.Stop == icetray.I3Frame.Physics:
            count.append(frame['Counter'].value)
```
